### Limpieza de Datos:

1. Valores negativos: le vamos a sacar el signo negativo a `total_venta`

2. Datos inconsistentes: le sacamos los títulos (Sr., Dr., etc.) que venían pegados al nombre `nombre_cliente`

3. Valores nulos:
 - `promo_producto`, `promo_pago`, `medio_pago` y `otros_productos`: vamos a completar los nulos con "sin promo", "sin especificar" y "ninguno", y a reemplazar los "???" de `promo_producto` por "sin promo"
 -  `precio_unitario` vamos a completarlo teniendo en cuenta la columna de `cantidad`y `total_venta`
 - `precio_unitario` y `cantidad`: las filas que tengan nulos en ambas no se pueden recuperar, así que las eliminamos (son 59)

4. Redonde de cantidad: pasamos a entero `cantidad`

5. Eliminar duplicados: eliminamos los duplicados `id_venta`, que resultaron ser errores de carga y no ventas repetidas. Una tiene el medio_pago real y la otra dice “sin especificar”

6. Creamos un nuevo ID




#### 1 - Valores Negativos

```python
df_ferreteria["total_venta"] = df_ferreteria["total_venta"].abs() # sacamos el signo negativo de la columna "total_venta" para que todos los valores sean positivos
```

```python
print("Valores negativos: {}".format((df_ferreteria["total_venta"] < 0).sum())) # verificamos que ya no hay valores negativos en la columna "total_venta"
```

#### 2 - Datos insonsistentes

```python
df_ferreteria["nombre_cliente"] = df_ferreteria["nombre_cliente"].str.replace(r"^\S+\.\s*", "", regex=True)
```

```python

print(df_ferreteria["nombre_cliente"].str.contains(r"\.", regex=True).sum()) # buscando puntos en los nombres vemos que ya no hay valores con el prefijo "Sr." o "Sra." en la columna "nombre_cliente"


df_ferreteria["nombre_cliente"].sample(10, random_state=2) # nos muestra 10 nombres de clientes aleatorios para verificar que se haya eliminado el prefijo
```

#### 3 - Valores nulos

```python
# 1. `promo_producto`, `promo_pago`, `otros_productos` y `medio_pago`: vamos a completar con las palabras "sin promo", "ninguno", "sin especificar"

df_ferreteria["promo_producto"] = df_ferreteria["promo_producto"].fillna("sin promo")
df_ferreteria["promo_producto"] = df_ferreteria["promo_producto"].replace("???", "sin promo")
df_ferreteria["promo_pago"] = df_ferreteria["promo_pago"].fillna("sin promo")
df_ferreteria["otros_productos"] = df_ferreteria["otros_productos"].fillna("ninguno")
df_ferreteria["medio_pago"] = df_ferreteria["medio_pago"].fillna("sin especificar")
```

```python
# Verificamos que las correcciones de los valores nulos se hayan realizado correctamente

print("Valores nulos en 'promo_producto':", df_ferreteria["promo_producto"].isnull().sum())
print("Valores '???' en 'promo_producto':", (df_ferreteria["promo_producto"] == "???").sum())

print("Valores nulos en 'promo_pago':", df_ferreteria["promo_pago"].isnull().sum())

print("Valores nulos en 'otros_productos':", df_ferreteria["otros_productos"].isnull().sum())

print("Valores nulos en 'medio_pago':", df_ferreteria["medio_pago"].isnull().sum())

```

```python
# Completamos solo los nulos

df_ferreteria["precio_unitario"] = df_ferreteria["precio_unitario"].fillna(df_ferreteria["total_venta"] / df_ferreteria["cantidad"])
df_ferreteria["cantidad"] = df_ferreteria["cantidad"].fillna(df_ferreteria["total_venta"] / df_ferreteria["precio_unitario"])
```

```python
print("Valores nulos en 'precio_unitario': {}".format(df_ferreteria["precio_unitario"].isnull().sum()))
print("Valores nulos en 'cantidad': {}".format(df_ferreteria["cantidad"].isnull().sum()))
```

```python
df_ferreteria["precio_unitario"] = df_ferreteria["precio_unitario"].round(2)
```

```python
# Corroboramos que ya no hay valores nulos en las columnas y celdas incompletas en el data

df_ferreteria.info()
```

#### 4 - Redondeo de Cantidad

```python
df_ferreteria["cantidad"] = df_ferreteria["cantidad"].round().astype(int)
```

```python
df_ferreteria.info()
```

#### 5 - Eliminamos duplicados del id_venta

```python
# Paso 1 (ordenar): Mandar las filas "sin especificar" al final, para que drop_duplicates se quede con la que tiene el dato real
df_ferreteria = df_ferreteria.sort_values("medio_pago", key=lambda s: s == "sin especificar", kind="stable")

# Paso 2. Borrar duplicados por id_venta, quedándose con la primera
df_ferreteria = df_ferreteria.drop_duplicates(subset="id_venta", keep="first")

# Paso 3. Reordenar por id_venta y resetear el índice
df_ferreteria = df_ferreteria.sort_values("id_venta").reset_index(drop=True)
```

```python
# Verificamos que ya no hay duplicados en la columna "id_venta" y que la cantidad de filas es la esperada

print("Total de filas:", len(df_ferreteria))
print("Duplicados en 'id_venta':", df_ferreteria["id_venta"].duplicated().sum())
print("Medio de pago para id_venta 1002:", df_ferreteria[df_ferreteria["id_venta"] == 1002]["medio_pago"].iloc[0]) # filtramos una fila por id_venta y mostramos el valor de la columna "medio_pago" para verificar que se haya quedado con el valor correcto
```

#### 6 - Creamos un nuevo id_cliente unico

```python
# Tenemos id que se duplicados por clientes distintos, vamos a ver cuantos hay

nombres_por_id = df_ferreteria.groupby("id_cliente")["nombre_cliente"].nunique()
print("Ids duplicados: {}".format((nombres_por_id > 1).sum()))
```

```python
nombres_por_id[nombres_por_id > 1].value_counts()

# 353 ids son compartidos por 2 personas, 43 por 3 y 3 por 4. Suman los 399
```

```python
# Un id por persona: mismo nombre = mismo id_cliente

df_ferreteria["id_cliente"] = df_ferreteria.groupby("nombre_cliente").ngroup() + 1
```

```python
print("Nombres distintos:", df_ferreteria["nombre_cliente"].nunique())
print("Ids distintos:", df_ferreteria["id_cliente"].nunique())
print("Clientes con más de una compra:", (df_ferreteria["id_cliente"].value_counts() > 1).sum())
```
